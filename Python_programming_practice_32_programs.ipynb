31)Write a program that checks if a number input by the user is a prime number using a for loop


```python
num=int(input("enter a number"))
if num>1:
    for i in range(2,num):
        if num%i==0:
            print("not a prime number")
            break
    else:
         print("Prime number")
else:
    print("Not a prime number")
```
